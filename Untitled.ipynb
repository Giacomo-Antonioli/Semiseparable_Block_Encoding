```python
from qiskit import QuantumCircuit
import numpy as np

# --- Setup ---
# Let's say we have 3 control qubits (0, 1, 2) and 1 target (3)
qc = QuantumCircuit(5) # 4 qubits for logic + 1 ancilla (qubit 4)

# --- Method 1: Arbitrary Phase (Ancilla Method) ---
theta = 0.3 # Example: 45 degrees

# 1. Compute: Flip ancilla (4) if all controls (0,1,2,3) are 1
# Note: In a CP gate, the "target" (3) also acts as a control for the phase condition
qc.mcx([0, 1, 2, 3], 4) 

# 2. Apply Phase to Ancilla
qc.p(theta, 4)

# 3. Uncompute: Reset ancilla to 0
qc.mcx([0, 1, 2, 3], 4)


qc.draw('mpl')
# Let's say we have 3 control qubits (0, 1, 2) and 1 target (3)
qcphase = QuantumCircuit(5) # 4 qubits for logic + 1 ancilla (qubit 4)

from qiskit.circuit.library import MCPhaseGate
qcphase.append(MCPhaseGate(theta,4),[0,1,2,3,4])
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator, state_fidelity, process_fidelity
from qiskit.circuit.library import MCPhaseGate
import numpy as np

# -----------------------
# PARAMETERS
# -----------------------
theta = 0.3

# -----------------------
# METHOD 1 — Manual Using MCX + P + MCX
# -----------------------
qc_manual = QuantumCircuit(5)  # qubits 0-3 logical, qubit 4 ancilla

# MCX controls = [0,1,2,3], target = 4
qc_manual.mcx([0, 1, 2, 3], 4)
qc_manual.p(theta, 4)
qc_manual.mcx([0, 1, 2, 3], 4)

# -----------------------
# METHOD 2 — Qiskit's Built-in MCPhaseGate
# -----------------------
qc_native = QuantumCircuit(5)
qc_native.append(MCPhaseGate(theta, num_ctrl_qubits=4), [0, 1, 2, 3, 4])

# -----------------------
# UNITARY MATRICES
# -----------------------
U_manual = Operator(qc_manual)
U_native = Operator(qc_native)

import numpy as np

print(np.linalg.norm(U_manual.data-U_native.data))
```
